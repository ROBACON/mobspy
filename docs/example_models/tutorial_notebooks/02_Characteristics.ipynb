# Characteristics

Meta-species in MobsPy can have characteristics (states). A `Bacteria` might be `.alive` or `.dead`. Characteristics partition a species into sub-populations, and reactions can target specific states.

```python
from mobspy import *
```

Add characteristics with dot notation. This creates a `Cell` species with states `alive` and `dead`.

```python
Cell = BaseSpecies()
Cell.alive, Cell.dead
```

Reactions can query specific states. Here, only alive cells die.

```python
Cell.alive >> Cell.dead @ 0.1

Cell.alive(100)
```

When you compile, MobsPy expands the meta-reaction into concrete reactions for each valid state combination.

```python
S = Simulation(Cell)
S.duration = 10
print(S.compile(verbose=True))
```

Characteristics form orthogonal spaces. If a species has two independent characteristic sets, MobsPy generates the Cartesian product.

```python
Age = BaseSpecies()
Age.young, Age.old

Health = BaseSpecies()
Health.healthy, Health.sick

Person = BaseSpecies(["Person"]) * Age * Health
Person.young.healthy(50)

S2 = Simulation(Person)
S2.duration = 10
print(S2.compile(verbose=True))
```

You can query multiple characteristics in a single reaction.

```python
Person.young.healthy >> Person.old.healthy @ 0.05

S3 = Simulation(Person)
S3.duration = 10
print(S3.compile(verbose=True))
```

Queries on products specify which state to transition TO. Unqueried characteristics stay unchanged (round-robin assignment from the reactant).
