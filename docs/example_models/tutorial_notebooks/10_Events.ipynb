# Events

Events change species counts at specific times or when conditions are met. They model interventions (adding a drug at t=10) or feedback (stop growth when population reaches a threshold).

```python
from mobspy import *
```

Time-triggered events use `event_time()` as a context manager. Inside the block, set species counts that will be applied at the specified time.

```python
A = BaseSpecies()

Zero >> A @ 1

A(10)

sim = Simulation(A)
sim.duration = 50
sim.plot_data = False

with sim.event_time(25):
    A(50)

sim.run()
print(sim.fres)
```

At t=25, the count of A jumps to 50 (overwriting whatever value it had grown to). The simulation then continues with the new count.

## Explicit Event API

The `sim.at()` and `sim.when()` methods are the preferred alternative to context managers. They take a dict mapping species to counts.

```python
A2 = BaseSpecies()

Zero >> A2 @ 1

A2(10)

sim_at = Simulation(A2)
sim_at.duration = 50
sim_at.plot_data = False

sim_at.at(25, {A2: 50})

sim_at.run()
print(sim_at.fres)
```

## Additive Events with Delta

By default, events *set* a species count to an absolute value. To *add* to the current count instead, use `Delta`. This is useful for modeling interventions like drug injections or immigration.

```python
from mobspy import Delta

A3 = BaseSpecies()

Zero >> A3 @ 1

A3(10)

sim_delta = Simulation(A3)
sim_delta.duration = 50
sim_delta.plot_data = False

# At t=25, ADD 100 to A3's count (not set to 100)
sim_delta.at(25, {A3: Delta(100)})

sim_delta.run()
print(sim_delta.fres)
```

`Delta(100)` adds 100 to the current count. Use `Delta(-50)` to subtract. Without `Delta`, the count is set to the exact value.

Condition-triggered events fire when a logical expression becomes true.

```python
B = BaseSpecies()

Zero >> B @ 5

B(0)

sim2 = Simulation(B)
sim2.duration = 100
sim2.plot_data = False

with sim2.event_condition(B >= 200):
    B(0)

sim2.run()
print(sim2.fres)
```

When B reaches 200, the event fires and resets B to 0. Growth then resumes from zero.

Logical operators combine conditions: `<=`, `>=`, `<`, `>`, `&` (and), `|` (or). Parentheses are required around each sub-condition when using `&` or `|`.

```python
C, D = BaseSpecies()

Zero >> C @ 3
Zero >> D @ 1

C(0)
D(0)

sim3 = Simulation(C | D)
sim3.duration = 100
sim3.plot_data = False

with sim3.event_condition((C >= 100) & (D >= 50)):
    C(0)
    D(0)

sim3.run()
print(sim3.fres)
```

Conditional duration: the simulation runs until a condition is met, rather than for a fixed time.

```python
E = BaseSpecies()

Zero >> E @ 10

E(0)

sim4 = Simulation(E)
sim4.duration = E >= 500
sim4.plot_data = False

sim4.run()
print(sim4.fres)
```

The simulation stops as soon as E reaches 500, regardless of the default time span.
