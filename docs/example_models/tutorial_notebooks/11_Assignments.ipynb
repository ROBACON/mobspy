# Assignments

Assignment rules define species whose counts are computed from other species at every time step. This is useful for derived quantities like total population, Hill function outputs, or ratio trackers.

```python
from mobspy import *
```

Use `.assign()` to make one species track a function of other species. At every time step, the assigned species' value is recomputed.

```python
A, B, Total = BaseSpecies()

Zero >> A[1]
Zero >> B[0.5]

with Assign:
    Total.assign(A + B)

A(10)
B(5)

sim = Simulation(A | B | Total)
sim.duration = 20
sim.plot_data = False

sim.run()
print(sim.fres)
```

Total tracks the sum of A and B at every point. It is not a real chemical species with reactions; its value is purely derived.

Hill functions are a common use case. Here, Signal follows a Michaelis-Menten curve as A grows.

```python
A2, Signal = BaseSpecies()

Zero >> A2[2]

with Assign:
    Signal.assign(A2 / (A2 + 100))

A2(0)

sim2 = Simulation(A2 | Signal)
sim2.duration = 100
sim2.plot_data = False

sim2.run()
print(sim2.fres)
```

As A2 grows from 0, Signal starts near 0 and saturates toward 1. The half-maximal response occurs when A2 reaches 100.

Assignments work with characteristic queries. You can assign based on a specific state of a species.

```python
Cell = BaseSpecies()
Cell.alive
Cell.dead

Tracker = BaseSpecies()

Cell.alive >> Cell.dead[0.01]

# Track a derived quantity: alive fraction
with Assign:
    Tracker.assign(Cell * 0.5)

Cell.alive(100)

sim3 = Simulation(Cell | Tracker)
sim3.duration = 50
sim3.plot_data = False

sim3.run()
print(sim3.fres)
```
