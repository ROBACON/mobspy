# Exporting Models

MobsPy can export compiled models to SBML (XML) and Antimony (human-readable) formats for use in COPASI, Tellurium, or other systems biology tools.

```python
from mobspy import *
```

Define a simple model to export.

```python
A, B = BaseSpecies()

A >> B @ 0.5
Zero >> A @ 1
B >> Zero @ 0.1

A(100)
B(0)

sim = Simulation(A | B)
sim.duration = 50
sim.compile()
```

`generate_sbml()` returns a list of SBML XML strings (one per simulation phase). The SBML is fully valid and can be loaded into any SBML-compatible tool.

```python
sbml_list = sim.generate_sbml()
print(sbml_list[0][:500])
```

`generate_antimony()` returns human-readable Antimony format. Antimony is a compact text representation of SBML models used by tools like Tellurium.

```python
antimony_list = sim.generate_antimony()
print(antimony_list[0])
```

When units are used, they appear in the exported model as SBML unit definitions.

```python
C, D = BaseSpecies()

C >> D @ (0.5 * (1 / u.second))
Zero >> C @ (1 * (1 / u.second))
D >> Zero @ (0.1 * (1 / u.second))

C(100)
D(0)

sim2 = Simulation(C | D)
sim2.duration = 50 * u.second
sim2.volume = 1 * u.liter
sim2.compile()

sbml_with_units = sim2.generate_sbml()
print(sbml_with_units[0][:800])
```
