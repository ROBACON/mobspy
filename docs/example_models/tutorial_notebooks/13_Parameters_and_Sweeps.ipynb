# Parameters and Sweeps

`ModelParameters` creates named parameters you can use in rates, then sweep over multiple values or update after compilation without recompiling.

```python
from mobspy import *
```

Create a parameter with `ModelParameters()` and use it in a rate expression. The parameter name is inferred from the variable name.

```python
A, B = BaseSpecies()

k = ModelParameters(0.1)

A >> B @ k

A(100)
B(0)

sim = Simulation(A | B)
sim.duration = 50
print(sim.compile(verbose=True))
```

The compiled output includes `k` as a named parameter with value 0.1. COPASI uses this parameter symbolically in the rate law.

After compilation, update parameters with `update_model()` and rerun without recompiling.

```python
sim.update_model((k, 0.5))
sim.plot_data = False
sim.run()
print("With k=0.5:")
print(sim.fres)
```

This is much faster than recompiling when you only need to adjust parameter values.

Sweep over parameter values by assigning a list. MobsPy runs one simulation per value and collects all results.

```python
C, D = BaseSpecies()

k2 = ModelParameters([0.01, 0.1, 1.0])

C >> D @ k2

C(100)
D(0)

sim2 = Simulation(C | D)
sim2.duration = 50
sim2.plot_data = False

sim2.run()
print(sim2.results)
```

The results contain three time series, one for each value of k2. This is the standard way to explore parameter sensitivity.

When multiple parameters are swept, MobsPy runs the **cartesian product** of all value combinations. For example, if `k1` has 3 values and `k2` has 4 values, you get 3 x 4 = 12 simulation runs. With `repetitions = N`, each combination is run N times, for a total of 3 x 4 x N runs.

Species initial counts can also be updated after compilation.

```python
sim.update_model((A, 200))
sim.run()
print("With A(200):")
print(sim.fres)
```
