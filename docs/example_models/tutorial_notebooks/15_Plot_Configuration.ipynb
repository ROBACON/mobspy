# Plot Configuration

MobsPy's plotting is built on matplotlib. You can customize colors, labels, line styles, and figure layout through `plot_config` or simulation parameters.

```python
from mobspy import *
```

Run a basic model to have data to plot.

```python
A, B, C = BaseSpecies()

A >> B @ 0.1
B >> C @ 0.05
C >> A @ 0.02

A(100)
B(50)
C(25)

sim = Simulation(A | B | C)
sim.duration = 100
sim.plot_data = False
sim.run()
```

Set global plot parameters via `plot_config`. These affect the overall figure appearance.

```python
sim.plot_config.xlabel = "Time (s)"
sim.plot_config.ylabel = "Count"
sim.plot_config.title = "Three-Species Cycle"

sim.plot_deterministic()
```

Customize individual species with `add_plot_params()`. Pass a dict mapping species to matplotlib style parameters.

```python
sim.add_plot_params(A, color="red", label="Species A", linestyle="--")
sim.add_plot_params(B, color="blue", label="Species B")

sim.plot_deterministic()
```

The `plot_raw()` method gives full control over figure layout when you need custom multi-panel figures.
