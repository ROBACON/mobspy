# Running Simulations

So far we have only compiled models. Now let's run them and look at the results. MobsPy supports deterministic (ODE) and stochastic (SSA) simulation via COPASI.

```python
from mobspy import *
```

## Deterministic Simulation

Define a simple production-and-decay model and run it deterministically. Species `A` is produced at rate 1 and decays at rate 0.1, starting from 100 molecules.

```python
A = BaseSpecies()

Zero >> A @ 1
A >> Zero @ 0.1

A(100)

sim = Simulation(A)
sim.duration = 50
sim.plot_data = False
sim.run()
```

## Accessing Results

After a run, use `sim.fres` (first result) to access time-series data. Index by the species object or by the string name.

```python
print("Time (first 5 points):", sim.fres["Time"][:5])
print("A (first 5 points):", sim.fres[A][:5])
```

## Stochastic Simulation

For stochastic simulation, set `simulation_method` to `'stochastic'` and run multiple repetitions. Each repetition uses a different random seed.

```python
A2 = BaseSpecies()

Zero >> A2 @ 1
A2 >> Zero @ 0.1

A2(100)

sim2 = Simulation(A2)
sim2.simulation_method = "stochastic"
sim2.repetitions = 3
sim2.duration = 50
sim2.run()
```

## Plotting Stochastic Results

For stochastic runs, `plot_stochastic()` shows the mean trajectory with standard deviation shading across repetitions.

```python
sim2.plot_stochastic()
```

## Key Simulation Parameters

| Parameter | Default | Description |
|---|---|---|
| `duration` | 60 | Simulation time length (can include units: `10 * u.hour`) |
| `simulation_method` | `'deterministic'` | `'deterministic'`, `'stochastic'`, `'hybrid'`, `'tauleap'` |
| `repetitions` | 1 | Number of stochastic runs |
| `volume` | 1 | Reaction volume (can include units: `1 * u.liter`) |
| `start_time` | 0 | Simulation start time |
| `plot_data` | True | Whether to display plots after running |
| `save_data` | False | Whether to save results to disk |
| `step_size` | None | Fixed output step size (None = auto) |

MobsPy writes the same rate laws for both deterministic and stochastic simulation. COPASI automatically handles the conversion from deterministic rate laws to stochastic propensities (e.g. converting `A * A * k` to `A * (A-1) / 2 * k` for the Gillespie algorithm).
