# Composing Simulations

Complex experiments often have phases: grow bacteria, add antibiotic, observe recovery. MobsPy lets you chain simulations with `+`, passing final counts from one as initial counts for the next.

```python
from mobspy import *
```

Define two simulation phases and compose them with `+`. The final species counts of phase 1 become the starting counts of phase 2.

```python
A, B = BaseSpecies()

# Phase 1: A grows
Zero >> A[5]
A(0)
B(0)

S1 = Simulation(A | B)
S1.duration = 20

# Phase 2: A converts to B
A >> B[0.1]

S2 = Simulation(A | B)
S2.duration = 30

# Compose
SC = S1 + S2
SC.plot_data = False
SC.run()
print(SC.fres)
```

The plot shows A growing during the first 20 time units, then converting to B in the next 30. The transition is seamless.

Each phase can have different simulation parameters (duration, method, volume).

```python
C, D = BaseSpecies()

Zero >> C[2]
C >> D[0.05]

C(10)
D(0)

S3 = Simulation(C | D)
S3.duration = 25
S3.simulation_method = "deterministic"

S4 = Simulation(C | D)
S4.duration = 25
S4.simulation_method = "stochastic"

SC2 = S3 + S4
SC2.plot_data = False
SC2.run()
print(SC2.fres)
```

The first phase runs deterministically (smooth curves), the second stochastically (noisy trajectories). This is useful for testing whether stochastic effects matter in a particular regime.

Events in composed simulations use local time. Each phase starts its clock at t=0.

```python
E = BaseSpecies()

Zero >> E[3]

E(0)

S5 = Simulation(E)
S5.duration = 20

S6 = Simulation(E)
S6.duration = 20

with S6.event_time(10):
    E(0)

SC3 = S5 + S6
SC3.plot_data = False
SC3.run()
print(SC3.fres)
```

The event at t=10 in phase 2 fires at absolute time 30 (20 from phase 1 + 10 from phase 2).
