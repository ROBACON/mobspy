# Inheritance

MobsPy's power comes from meta-reactions that apply to entire families of species. When you define a reaction on a parent, all children inherit it. This lets you write one reaction that expands to many.

```python
from mobspy import *
```

Create child species with `New()`. Children inherit all reactions from their parent.

```python
Animal = BaseSpecies()
Cat, Dog = New(Animal, ["Cat", "Dog"])

# All animals die
Animal >> Zero @ 0.01

Cat(50)
Dog(50)

S = Simulation(Animal)
S.duration = 10
print(S.compile(verbose=True))
```

Both Cat and Dog inherited the death reaction from Animal. Children can have their own reactions too.

```python
# Only cats reproduce
Cat >> Cat + Cat @ 0.1

S2 = Simulation(Animal)
S2.duration = 10
print(S2.compile(verbose=True))
```

Cat now has both death and reproduction, while Dog only has death.

The `*` operator creates a child that combines characteristics from another species. This is how you add state spaces to existing species.

```python
Base = BaseSpecies()
Age = BaseSpecies()
Age.young, Age.old

Organism = Base * Age
Organism.young(100)

S3 = Simulation(Organism)
S3.duration = 10
print(S3.compile(verbose=True))
```

The `*` operator merges the characteristic spaces. `Organism` now has the `young`/`old` states from `Age`.
