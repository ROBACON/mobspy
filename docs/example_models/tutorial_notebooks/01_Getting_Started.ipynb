# Getting Started

MobsPy lets you define chemical reaction networks using Python operators. This notebook covers the basics: creating species, writing reactions, and compiling a model.

```python
from mobspy import *
```

Species are the building blocks. Create them with `BaseSpecies()`, passing names explicitly.

```python
A, B = BaseSpecies()
```

Reactions use `>>` for transformation and `@` for rates. This says A converts to B at rate 1.

```python
A >> B @ 1
```

Use `Zero` for creation (birth) and destruction (death) reactions.

```python
Zero >> A @ 0.5
B >> Zero @ 0.1
```

Set initial counts with the call operator `()`.

```python
A(100)
B(0)
```

Create a `Simulation`, compile, and inspect the result.

```python
S = Simulation(A | B)
S.duration = 10
result = S.compile(verbose=True)
print(result)
```

The compiled output shows the species, reactions, and parameters MobsPy generated. Each reaction has reactants, products, and a kinetic rate law.

Here is a slightly larger example: a Lotka-Volterra predator-prey model with three reactions.

```python
Predator, Prey = BaseSpecies()

# Prey reproduces
Zero >> Prey @ 1.0

# Predation: predator eats prey and reproduces
Predator + Prey >> Predator + Predator @ 0.01

# Predator dies
Predator >> Zero @ 0.1

Predator(10)
Prey(100)

S2 = Simulation(Predator | Prey)
S2.duration = 50
print(S2.compile(verbose=True))
```
