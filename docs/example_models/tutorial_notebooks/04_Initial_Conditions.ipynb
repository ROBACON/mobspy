# Initial Conditions

Before simulating, you need to specify how many of each species exist at time zero. MobsPy provides several ways to set initial counts, including queries for specific states and the `All` operator for bulk assignment.

```python
from mobspy import *
```

The simplest case: set a count on a species with no characteristics.

```python
A = BaseSpecies()
A(100)

S = Simulation(A)
S.duration = 1
print(S.compile(verbose=True))
```

For species with characteristics, counts are assigned to specific states via queries.

```python
Color = BaseSpecies()
Color.red, Color.blue

Ball = BaseSpecies(["Ball"]) * Color
Ball.red(30)
Ball.blue(70)

S2 = Simulation(Ball)
S2.duration = 1
print(S2.compile(verbose=True))
```

States you don't assign get count 0 by default. When no query is given, the count goes to the default state (the first characteristic listed for each axis).

```python
Particle = BaseSpecies()
Particle.x, Particle.y, Particle.z

# This assigns 50 to the default state (Particle.x)
Particle(50)

S3 = Simulation(Particle)
S3.duration = 1
print(S3.compile(verbose=True))
```

`All[]` assigns a count to every state of a species.

```python
Token = BaseSpecies()
Token.a, Token.b, Token.c

All[Token](10)

S4 = Simulation(Token)
S4.duration = 1
print(S4.compile(verbose=True))
```

`All[]` also accepts queries to target a subset of states.

```python
Shape = BaseSpecies()
Shape.red, Shape.blue

Size = BaseSpecies()
Size.small, Size.big

Obj = BaseSpecies(["Obj"]) * Shape * Size

# Only red objects get count 5 (both red.small and red.big)
All[Obj.red](5)

S5 = Simulation(Obj)
S5.duration = 1
print(S5.compile(verbose=True))
```

`set_counts()` is an alternative syntax for setting multiple counts at once.

```python
X = BaseSpecies()
X.on, X.off
Y = BaseSpecies()

set_counts({X.on: 40, X.off: 60, Y: 200})

S6 = Simulation(X | Y)
S6.duration = 1
print(S6.compile(verbose=True))
```

Born species: when a product in a reaction has no matching reactant, MobsPy assigns it the default state. Use `All[]` in the product to create all states instead.

```python
Src = BaseSpecies()
Dst = BaseSpecies()
Dst.alpha, Dst.beta

# Dst gets its default state (alpha) in the product
Src >> Dst @ 1

Src(10)

S7 = Simulation(Src | Dst)
S7.duration = 1
print(S7.compile(verbose=True))
```
